```python
def rectangle_area(length, width):
    return length * width
print(rectangle_area(2,4))
```

```python
def is_even(n):
    return n%2==0
print(is_even(8))
```

```python
def sum_list(numbers):
    return sum(numbers)
print(sum_list([9,8,7,3]))
```

```python
def reverse_string(s):
    return s[::-1]
print(reverse_string("hey buddy"))
```

```python
def find_max(numbers):
    return max(numbers)
print(find_max([29, 100, 2, 10, 9]))
```

```python
def square(n):
    return n**2
print(square(6))
```

```python
def is_palindrome(a):
    return a==a[::-1]
print(is_palindrome("dad"))
```

```python
def count_vowels(s):
    count = 0
    for char in s:
        if char in "aeiou":
            count+=1
    return count
print(count_vowels("hello aksh"))
```

```python
def factorial(n):
    result=1
    for i in range(1, n + 1):
        result = result*i
    return result
print(factorial(6))
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n%i==0:
            return False
    return True
print(is_prime(5))
```

```python
def celsius_to_fahrenheit(c):
    return c * 9/5 + 32
print(celsius_to_fahrenheit(33))
```

```python
def count_occurrences(lst,item):
    return lst.count(item)
print(count_occurrences("item","honey,sugar,cheese"))
```
